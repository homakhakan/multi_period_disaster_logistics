```python
#pip install matplotlib
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib.ticker import FormatStrFormatter
```

```python
xls = pd.ExcelFile('Sensitivity Analysis.xlsx')
df1 = pd.read_excel(xls, 'Weights of Objs')
df2 = pd.read_excel(xls, 'All Cots - Weights of Objs')
df3 = pd.read_excel(xls, '# of Trucks Obj')
df4 = pd.read_excel(xls, '# of Trucks PC')
df5 = pd.read_excel(xls, '# of Trucks Cost')
df6 = pd.read_excel(xls, 'All Cots - # of Trucks')
df6.head
```

```python
x = df1['Penalty Cost'][2:9]
y = df1['Financial Cost'][2:9]
labels = df1['wᵇ'][2:9].astype(str)

# Create scatter plot with custom palette
plt.figure(figsize=(10, 7))
sns.scatterplot(x=x, y=y, hue=labels, palette="husl", s=100)

# Add a legend
plt.legend(fontsize='large')

# Set axis labels
plt.xlabel('Penalty Cost', fontsize='large')
plt.ylabel('Financial Cost', fontsize='large')
plt.gca().yaxis.set_major_formatter(FormatStrFormatter('$%d'))
plt.gca().xaxis.set_major_formatter(FormatStrFormatter('%d'))

# Set a title
plt.title('Effect of Various Weights for Financial Cost (wᵇ) on Objectives\n'
          'wᵇ: 1/100, 1/10, 1/8, 1/6, 1/4, 1/2, 1     # Trucks=20', fontsize='large')

# Display the plot
plt.show()
```

```python
x = [14,16,18,20,22,24]
y1 = df3['Mix Trucks'][0:6]
y2 = df3["Single Comm."][0:6]

plt.figure(figsize=(10, 7))

plt.plot(x, y1, label='Mixed Load')
plt.plot(x, y2, label='Single-Commodity')

# Add a legend
plt.legend(fontsize='large')

# Set axis labels
plt.xlabel('Number of Trucks', fontsize='large')
plt.ylabel('Total Objective Value', fontsize='large')
plt.gca().yaxis.set_major_formatter(FormatStrFormatter('%d'))

# Set a title
plt.title('Objective Value: Sensitivity Analysis on Number of Trucks\n'
          'nT= 14, 16, 18, 20, 22, 24', fontsize=14)

# Display the plot
plt.show()
```
