# Boxplot

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import ScalarFormatter

# Load the Excel file (update with the actual file path)
file_path = 'boxplot.xlsx'  # Replace with your actual file path
data = pd.read_excel(file_path)

# Check the column names
print(data.columns)

# Extract the relevant columns for the box plot
columns_of_interest = ['20 scenarios', '30 scenarios', '50 scenarios']

# Convert columns to numeric (to handle any non-numeric values)
data_cleaned = data[columns_of_interest].apply(pd.to_numeric, errors='coerce')

# Create a box plot with black color, narrower boxes, and mean markers
plt.figure(figsize=(5, 3))
plt.boxplot(
    [data_cleaned[col] for col in columns_of_interest], 
    labels=columns_of_interest, 
    patch_artist=True,
    widths=0.3,
    boxprops=dict(facecolor='white', color='black'),  # Box border and fill color
    whiskerprops=dict(color='black'),                 # Whisker color
    capprops=dict(color='black'),                     # Cap color
    flierprops=dict(markerfacecolor='black', color='black'),  # Outlier color
    medianprops=dict(color='black'),                  # Median line color
    meanprops=dict(marker='x', markerfacecolor='black', markeredgecolor='black'),  # Mean marker
    showmeans=True                                    # Enable mean markers
)

# Use ScalarFormatter to show actual values on the y-axis without scientific notation
ax = plt.gca()
formatter = ScalarFormatter(useOffset=False)
formatter.set_scientific(False)
ax.yaxis.set_major_formatter(formatter)

# Add horizontal grid only
plt.grid(axis='y')

# Save the plot to a file
plt.savefig('boxplot_output1.png', dpi=300, bbox_inches='tight')  # Save as PNG with high resolution

# Display the plot
plt.show()
```
