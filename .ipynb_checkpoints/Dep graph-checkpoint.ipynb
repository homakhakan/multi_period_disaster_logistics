```python
import numpy as np
import matplotlib.pyplot as plt

# Configuration labels
configs = ['(0.2,1)', '(5,1)', '(1,1)', '(1,0.2)', '(1,5)']

# Percentages per bucket (rows = configs, columns = buckets)
# Order: Bucket 1, 2, 3, 4, 5+
data_small = np.array([
    [69.0, 24.3, 4.9, 1.9, 0.0],
    [67.7, 19.9, 7.9, 4.5, 0.0],
    [71.0, 21.5, 5.5, 2.0, 0.0],
    [69.5, 20.9, 6.2, 3.3, 0.0],
    [67.5, 23.9, 6.1, 2.4, 0.0]
])

bucket_labels = ['B1', 'B2', 'B3', 'B4', 'B5+']
colors = ['#4C72B0', '#55A868', '#C44E52', '#8172B2', '#CCB974']

fig, ax = plt.subplots(figsize=(4, 4))

bottom = np.zeros(len(configs))

for i in range(data_small.shape[1]):
    ax.bar(
        configs,
        data_small[:, i],
        bottom=bottom,
        label=bucket_labels[i],
        color=colors[i],
        edgecolor='black',
        linewidth=0.5,
        width=0.5
    )
    bottom += data_small[:, i]

ax.set_ylabel('Percentage of Unmet Demand')
ax.set_xlabel('(Penalty Cost Factor, Transportation Cost Factor)')
ax.set_title('Distribution of Unmet Demand Across Deprivation Buckets\n(Small Network)')
ax.set_ylim(0, 100)
ax.legend(
    title='Buckets',
    loc='lower right',
    fontsize=6,
    title_fontsize=7
)

plt.tight_layout()
plt.savefig('unmet_bucket_distribution_small_network.png', dpi=300)
plt.show()


import numpy as np
import matplotlib.pyplot as plt

# Configuration labels
configs = ['(0.2,1)', '(5,1)', '(1,1)', '(1,0.2)', '(1,5)']

# Percentages per bucket (rows = configs, columns = buckets)
# Order: Bucket 1, 2, 3, 4, 5+
data_large = np.array([
    [57.5, 26.7, 12.1, 3.7, 0.0],
    [76.1, 19.4, 3.5, 0.9, 0],
    [69.8, 22.4, 6.3, 1.6, 0.0],
    [78.0, 16.4, 4.9, 0.6, 0.0],
    [57.7, 26.9, 11.6, 3.8, 0.0]
])

bucket_labels = ['B1', 'B2', 'B3', 'B4', 'B5+']
colors = ['#4C72B0', '#55A868', '#C44E52', '#8172B2', '#CCB974']

fig, ax = plt.subplots(figsize=(4, 4))

bottom = np.zeros(len(configs))

for i in range(data_large.shape[1]):
    ax.bar(
        configs,
        data_large[:, i],
        bottom=bottom,
        label=bucket_labels[i],
        color=colors[i],
        edgecolor='black',
        linewidth=0.5,
        width=0.5
    )
    bottom += data_large[:, i]

ax.set_ylabel('Percentage of Unmet Demand')
ax.set_xlabel('(Penalty Cost Factor, Transportation Cost Factor)')
ax.set_title('Distribution of Unmet Demand Across Deprivation Buckets\n(Large Network)')
ax.set_ylim(0, 100)
ax.legend(
    title='Buckets',
    loc='lower right',
    fontsize=6,
    title_fontsize=7
)

plt.tight_layout()
plt.savefig('unmet_bucket_distribution_large_network.png', dpi=300)
plt.show()
```
